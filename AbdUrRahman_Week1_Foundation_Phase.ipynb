```python
# MINI PROJECT: STUDENT MARKS MANAGEMENT SYSTEM

class StudentMarks:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_average(self):
        return sum(self.marks) / len(self.marks)

    def get_grade(self):

        average = self.calculate_average()

        if average >= 90:
            return "A"
        elif average >= 80:
            return "B"
        elif average >= 70:
            return "C"
        elif average >= 60:
            return "D"
        else:
            return "F"

    def display_result(self):

        print("Student:", self.name)
        print("Marks:", self.marks)
        print("Average:", round(self.calculate_average(), 2))
        print("Grade:", self.get_grade())


student = StudentMarks(
    "Abd Ur Rahman",
    [82, 76, 91, 88, 79]
)

student.display_result()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
df.describe(include="all").T
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nMissing percentage:")
print((df.isnull().mean() * 100).round(2))
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Create a copy so the original dataset remains unchanged

cleaned_df = df.copy()

# Fill missing Age values using the median
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())

# Fill missing Embarked values using the mode
cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(
    cleaned_df["Embarked"].mode()[0]
)

# Remove Cabin because it contains a large number of missing values
cleaned_df = cleaned_df.drop(columns=["Cabin"])

print("Missing values after cleaning:")
print(cleaned_df.isnull().sum())
```

```python
print("Missing values remaining:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows:")
print(cleaned_df.duplicated().sum())
```

```python
survival_counts = cleaned_df["Survived"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(
    ["Did Not Survive", "Survived"],
    survival_counts.values
)

plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    cleaned_df["Age"],
    bins=20
)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    cleaned_df["Fare"],
    bins=30
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
gender_survival = pd.crosstab(
    cleaned_df["Sex"],
    cleaned_df["Survived"]
)

gender_survival.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=0)
plt.legend(["Did Not Survive", "Survived"])
plt.show()
```

```python
class_survival = pd.crosstab(
    cleaned_df["Pclass"],
    cleaned_df["Survived"]
)

class_survival.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=0)
plt.legend(["Did Not Survive", "Survived"])
plt.show()
```

```python
average_fare = cleaned_df.groupby("Pclass")["Fare"].mean()

print(average_fare)

plt.figure(figsize=(7, 5))

plt.bar(
    average_fare.index.astype(str),
    average_fare.values
)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

```python
numeric_columns = cleaned_df.select_dtypes(
    include=np.number
)

correlation = numeric_columns.corr()

print(correlation)
```

```python
plt.figure(figsize=(10, 7))

plt.imshow(correlation, aspect="auto")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# TASK 1.4 - TWO ML MODELS WITH SCIKIT-LEARN

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Libraries imported successfully.")
```

```python
housing = fetch_california_housing(
    as_frame=True
)

X_regression = housing.data
y_regression = housing.target

print("Features:")
display(X_regression.head())

print("\nTarget:")
display(y_regression.head())

print("\nDataset shape:")
print(X_regression.shape)
```

```python
print("Missing values:")
print(X_regression.isnull().sum())

print("\nFeature information:")
print(X_regression.info())
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_regression,
    y_regression,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train_reg))
print("Testing samples:", len(X_test_reg))
```

```python
regression_scaler = StandardScaler()

X_train_reg_scaled = regression_scaler.fit_transform(
    X_train_reg
)

X_test_reg_scaled = regression_scaler.transform(
    X_test_reg
)

print("Regression feature scaling completed.")
```

```python
regression_model = LinearRegression()

regression_model.fit(
    X_train_reg_scaled,
    y_train_reg
)

print("Linear Regression model trained successfully.")
```

```python
regression_predictions = regression_model.predict(
    X_test_reg_scaled
)

print("First 10 predictions:")
print(regression_predictions[:10])
```

```python
mae = mean_absolute_error(
    y_test_reg,
    regression_predictions
)

mse = mean_squared_error(
    y_test_reg,
    regression_predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test_reg,
    regression_predictions
)

print("LINEAR REGRESSION EVALUATION")

print("MAE :", round(mae, 4))
print("MSE :", round(mse, 4))
print("RMSE:", round(rmse, 4))
print("R²  :", round(r2, 4))
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_reg,
    regression_predictions,
    alpha=0.5
)

plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")

plt.title(
    "Actual vs Predicted House Values"
)

plt.show()
```

```python
from sklearn.datasets import load_breast_cancer

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

print("Classification libraries loaded.")
```

```python
cancer = load_breast_cancer(
    as_frame=True
)

X_classification = cancer.data
y_classification = cancer.target

print("Dataset shape:")
print(X_classification.shape)

print("\nFeatures:")
display(X_classification.head())

print("\nTarget:")
display(y_classification.head())
```

```python
print(
    "Total missing values:",
    X_classification.isnull().sum().sum()
)
```

```python
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_classification,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)

print("Training samples:", len(X_train_class))
print("Testing samples:", len(X_test_class))
```

```python
classification_scaler = StandardScaler()

X_train_class_scaled = classification_scaler.fit_transform(
    X_train_class
)

X_test_class_scaled = classification_scaler.transform(
    X_test_class
)

print("Classification feature scaling completed.")
```

```python
classification_model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

classification_model.fit(
    X_train_class_scaled,
    y_train_class
)

print("Logistic Regression model trained successfully.")
```

```python
classification_predictions = classification_model.predict(
    X_test_class_scaled
)

print("First 20 predictions:")
print(classification_predictions[:20])
```

```python
accuracy = accuracy_score(
    y_test_class,
    classification_predictions
)

precision = precision_score(
    y_test_class,
    classification_predictions
)

recall = recall_score(
    y_test_class,
    classification_predictions
)

f1 = f1_score(
    y_test_class,
    classification_predictions
)

print("LOGISTIC REGRESSION EVALUATION")

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
```

```python
print(
    classification_report(
        y_test_class,
        classification_predictions,
        target_names=cancer.target_names
    )
)
```

```python
cm = confusion_matrix(
    y_test_class,
    classification_predictions
)

print("Confusion Matrix:")
print(cm)

display_object = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=cancer.target_names
)

display_object.plot()

plt.title(
    "Confusion Matrix - Logistic Regression"
)

plt.show()
```
